## Задание 3

1. В датафрейме transactions найдите клиента с максимальной суммой приходов на карту
2. В датафрейме transactions найдите клиента с максимальной суммой расходов по карте
3. Найдите модуль разницы для этих клиентов между суммой расходов и суммой приходов

```python
income_df = df[df['amount'] > 0]
customer_income = income_df.groupby('customer_id')['amount'].sum()
max_income_customer = customer_income.idxmax()
max_income_sum = customer_income.max()

print(f"Клиент с максимальной суммой приходов: {max_income_customer}, сумма: {max_income_sum:.2f}")
```

```python
expense_df = df[df['amount'] < 0]
customer_expense = expense_df.groupby('customer_id')['amount'].sum().abs()
max_expense_customer = customer_expense.idxmax()
max_expense_sum = customer_expense.max()

print(f"Клиент с максимальной суммой расходов: {max_expense_customer}, сумма: {max_expense_sum:.2f}")
```

```python
income_customer_income = df[(df['customer_id'] == max_income_customer) & (df['amount'] > 0)]['amount'].sum()
income_customer_expense = df[(df['customer_id'] == max_income_customer) & (df['amount'] < 0)]['amount'].sum()
diff_for_income_customer = abs(income_customer_expense - income_customer_income)

print(f"Модуль разницы (расходы - приходы) для клиента с максимальной суммой приходов {max_expense_customer}: {diff_for_income_customer:.2f}")
```

```python
expense_customer_income = df[(df['customer_id'] == max_expense_customer) & (df['amount'] > 0)]['amount'].sum()
expense_customer_expense = df[(df['customer_id'] == max_expense_customer) & (df['amount'] < 0)]['amount'].sum()
diff_for_expense_customer = abs(expense_customer_expense - expense_customer_income)

print(f"Модуль разницы (расходы - приходы) для клиента с максимальной суммой расходов {max_expense_customer}: {diff_for_expense_customer:.2f}")
```

Получилось одинаково, т.к. клиент с максимальной суммой расходов и приходов один и тот же

## Задание 5
1. Определите модуль разницы между средними тратами женщин и мужчин (трата - отрицательное значение amount).
2. Определите модуль разницы между средними поступлениями у мужчин и женщин

```python
expenses = transactions[transactions['amount'] < 0].copy()
expenses['expense_amount'] = expenses['amount'].abs()

avg_expense_by_gender = expenses.groupby('gender')['expense_amount'].mean()
print(f"Средние траты: {avg_expense_by_gender}")

diff_expenses = abs(avg_expense_by_gender.iloc[0] - avg_expense_by_gender.iloc[1])
print(f"Модуль разницы средних трат: {diff_expenses:.2f}")
```

```python
incomes  = transactions[transactions['amount'] > 0].copy()
incomes ['incomes_amount'] = expenses['amount'].abs()

avg_income_by_gender = incomes.groupby('gender')['amount'].mean()
print(f"Средние траты: {avg_income_by_gender}")

diff_incomes = abs(avg_income_by_gender.iloc[0] - avg_income_by_gender.iloc[1])
print(f"Модуль разницы средних поступлений: {diff_incomes:.2f}")
```

## Задание 6
1. По всем типам транзакций рассчитайте максимальную сумму прихода на карту (из строго положительных сумм по столбцу amount) отдельно для мужчин и женщин (назовите ее "max_income"). Оставьте по 10 типов транзакций для мужчин и для женщин, наименьших среди всех типов транзакций по полученным значениям "max_income".
2. Выделите среди них те типы транзакций, которые встречаются одновременно и у мужчин, и у женщин

```python
max_income_by_gender_type = incomes.groupby(['gender', 'tr_type'])['amount'].max().reset_index()
max_income_by_gender_type.columns = ['gender', 'tr_type', 'max_income']

men_max_income = max_income_by_gender_type[max_income_by_gender_type['gender'] == 1].nsmallest(10, 'max_income')
women_max_income = max_income_by_gender_type[max_income_by_gender_type['gender'] == 0].nsmallest(10, 'max_income')

print("Топ-10 типов транзакций с наименьшим max_income для мужчин:")
print(men_max_income[['tr_type', 'max_income']])
print("\nТоп-10 типов транзакций с наименьшим max_income для женщин:")
print(women_max_income[['tr_type', 'max_income']])
```

## Задание 8
1. Из поля tr_datetime выделите час tr_hour, в который произошла транзакция, как первые 2 цифры до ":". (**)
2. Посчитайте количество транзакций с amount<0 в ночное время для мужчин и женщин. Ночное время - это примерно 00-06 часов.

```python
transactions['tr_hour'] = transactions['tr_datetime'].str.split().str[1].str.split(':').str[0].astype(int)
print(transactions[['tr_datetime', 'tr_hour']].head())
```

```python
night_transactions = transactions[(transactions['amount'] < 0) & (transactions['tr_hour'].between(0, 6))]

night_count_by_gender = night_transactions.groupby('gender').size()

print(f"Количество ночных трат (00-06) для женщин (gender=0): {night_count_by_gender.get(0, 0)}")
print(f"Количество ночных трат (00-06) для мужчин (gender=1): {night_count_by_gender.get(1, 0)}")
print(f"Всего ночных трат: {night_count_by_gender.sum()}")
```
